**Program to implement linear and multiple regression techniques using standard dataset available in the public domain and evaluate its performance**

```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn import datasets, linear_model
from sklearn.metrics import mean_squared_error, r2_score
```

```python
df=datasets.load_diabetes()
df['feature_names']
```

```python
diabetes_X, diabetes_y = datasets.load_diabetes(return_X_y=True)
diabetes_X.shape
```

```python
diabetes_y.shape
```

```python
diabetes_X = diabetes_X[:, np.newaxis,2]
diabetes_X.shape
```

```python
diabetes_X_train = diabetes_X[:-20]
diabetes_X_test = diabetes_X[-20:]

diabetes_y_train = diabetes_y[:-20]
diabetes_y_test = diabetes_y[-20:]
```

```python
regr = linear_model.LinearRegression()
regr.fit(diabetes_X_train, diabetes_y_train)
```

```python
diabetes_y_pred = regr.predict(diabetes_X_test)
```

```python
print("Coefficients: \n", regr.coef_)
print("Mean squared error:%.2f" %mean_squared_error(diabetes_y_test,diabetes_y_pred))
print("coefficient of determination: %.2f" % r2_score(diabetes_y_test,diabetes_y_pred))
```

```python
plt.scatter(diabetes_X_test,diabetes_y_test,color="black")
plt.plot(diabetes_X_test,diabetes_y_pred,color="blue",linewidth=3)
plt.xlabel("age")
plt.ylabel("diabetes progression")
plt.xticks(())
plt.yticks(())
plt.show()
```

**MULTIPLE REGRESSION**

```python
diabetes_X,diabetes_y = datasets.load_diabetes(return_X_y=True)
diabetes_X.shape
diabetes_X=diabetes_X[:,[0,2]]
diabetes_X.shape
```

```python
diabetes_X_train = diabetes_X[:-20]
diabetes_X_test = diabetes_X[-20:]

diabetes_y_train = diabetes_y[:-20]
diabetes_y_test = diabetes_y[-20:]
```

```python
regr = linear_model.LinearRegression()
regr.fit(diabetes_X_train,diabetes_y_train)
```

```python
diabetes_y_pred = regr.predict(diabetes_X_test)
```

```python
print("Coefficients:\n", regr.coef_)
print("Intercept: \n", regr.intercept_)

print("Mean squared error: %.2f" % mean_squared_error(diabetes_y_test, diabetes_y_pred))
print("Coefficient of determination: %.2f" % r2_score(diabetes_y_test, diabetes_y_pred))
```

```python
x = diabetes_X_test[:,0]
y = diabetes_X_test[:,1]
#z = diabetes_X_test[:,2]
```

```python
plt.style.use('default')

fig = plt.figure(figsize=(12, 4))

ax1 = fig.add_subplot(131, projection='3d')
ax2 = fig.add_subplot(132, projection='3d')
ax3 = fig.add_subplot(133, projection='3d')

axes = [ax1, ax2, ax3]

for ax in axes:
    ax.plot(
        x, y, diabetes_y_pred,
        color='k',
        zorder=15,
        linestyle='none',
        marker='o',
        alpha=0.5
    )

    ax.scatter(
        x.flatten(),
        y.flatten(),
        diabetes_y_pred,
        facecolor=(0, 0, 0, 0),
        s=20,
        edgecolor='#70b3f0'
    )

    ax.set_xlabel('Age', fontsize=12)
    ax.set_ylabel('BMI', fontsize=12)
    ax.set_zlabel('diabetes', fontsize=12)

    ax.locator_params(nbins=4, axis='x')
    ax.locator_params(nbins=5, axis='y')

ax1.view_init(elev=28, azim=120)
ax2.view_init(elev=4, azim=114)
ax3.view_init(elev=60, azim=165)

fig.suptitle(
    f'R2 = {r2_score(diabetes_y_test, diabetes_y_pred):.3f}',
    fontsize=20
)

fig.tight_layout()
```
