**Write a pgm to draw a scatter plot comparing two subject marks of mathematics and science**

```python
import matplotlib.pyplot as plt

x = [10, 20, 30, 40, 50, 60, 70, 80, 100]
m = [88, 92, 80, 89, 100, 80, 60, 100, 80]
s = [35, 79, 79, 48, 100, 88, 32, 45, 20]

plt.scatter(x, m, label="Maths Marks")
plt.scatter(x, s, label="Science Marks")

plt.legend(loc='upper right')
plt.xlabel("Marks Range")
plt.ylabel("Marks Scored")
plt.title("Scatter Plot")

plt.show()


```

**Write a pgm to create bar plot of scores by group and gender.use multiple X values on the same chart for men and women**

```python
import numpy as np
import matplotlib.pyplot as plt
y1 = [22,30,35,35,26]
y2 = [25,32,30,35,29]
x_labels =['G1','G2','G3','G4','G5']
x1 = np.arange(5)
width = 0.40
plt.bar(x1-0.2,y1,color="green",width=width,label='Men')
plt.bar(x1+0.2,y2,color="red",width=width,label='Women')
plt.xticks(x1,x_labels)
plt.ylabel("Scores")
plt.legend()
plt.title("Scores by group and gender")
plt.show()
```

**Write a pgm to create a pie chart of the popularity of programming languages**

```python
import matplotlib.pyplot as plt
import numpy as np

y = np.array([22.2, 17.6, 8.8, 7.7, 6.7, 5.0])
mylabels = ["Java", "Python", "PHP", "JavaScript", "C#", "C++"]

plt.pie(y, labels=mylabels)
plt.show()
```
